### 6. Data Formatting & Type Conversion
#### Convert Species to Categorical Type

```python
df['species'] = df['species'].astype('category')
```

### 7. Data Normalization

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df[['sepal_length','sepal_width','petal_length','petal_width']] = scaler.fit_transform(
    df[['sepal_length','sepal_width','petal_length','petal_width']]
)
```

### 8. Convert Categorical Variable into Quantitative Variable
#### Label Encoding Species Column

```python
df['species'] = df['species'].cat.codes
```
